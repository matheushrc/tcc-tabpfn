```python
import missingno as msno
import pandas as pd

%matplotlib inline
```

```python
mines = pd.read_csv(
    "SACM/SACM_DR_csv/Geology.csv",
    encoding="cp1252",
    low_memory=False,
)
```

```python
mines_cleaned = mines.drop(
    columns=[
        "SAMPLE_COMMENT_ST",
        "SAMPLE_COMMENT_LT",
        "PROJECT_NAME",
        "DATE_SUBMITTED",
        "DATE_COLLECTED",
        "DISTRICT_NAME",
        "MINE_NAME",
        "REF_SHORT_NAME",
        "DEPOSIT_NAME",
        "SUBMITTER",
        "COORDINATES_COMMENT",
        "LOCATE_DESC",
        "HOST_NAME",
        "LAB_ID",
        "PREV_LAB_ID_1",
        "JOB_ID",
        "PREV_JOB_ID_1",
    ],
    index=None,
)

mines_cleaned.to_csv("output/cleaned/Geology.csv", index=None)
```
